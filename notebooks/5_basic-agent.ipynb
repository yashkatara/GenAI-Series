```python
from  dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.tools import tool
```

```python
@tool
def add_number (a:int,b:int):
  """
  it will return the sum of two numbers
  Args: 
      a: Number one
      b:Number two
  """
  return a+b

@tool
def multiply_number (a:int,b:int):
  """
  it will return the sum of two numbers
  Args: 
      a: Number one
      b:Number two
  """
  return a*b
```

```python
add_number.invoke({'a':12, 'b':3})
```

```python
from langchain_openai import ChatOpenAI
from langchain.agents import create_agent
```

```python
llm =  ChatOpenAI(model='gpt-4')
```

```python
agent = create_agent(
  model = llm,
  tools=[add_number,multiply_number],
  system_prompt = " You are a math teacher and always use tool for calculation"
)
```

```python
response = agent.invoke({'messsages':[{'role':'user' , 'content':'what is 2+3'}]})
```

```python
print(response['messages'][-1].content)
```

```python
for res in response['messages']:
  print(res)
  print('\n')
```
